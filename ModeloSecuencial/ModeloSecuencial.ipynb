### Modelo Secuencial 
Es una arquitectura en la que las capas se apilan de forma lineal, una tras otra, creando una secuencia simple desde la entrada a la salida, cada capa tiene exactamente un tensor de entrada y un tensro de salida 

Es de un enfoque sencillo y organizado para construir redes neuronales, en donde se carateriza por la unica direccion en donde se dirige la informacion 

No es adecuad para aquitecturas complejas que requieren multiples entradas, multiples salidas, capas compartidas o ramificaciones, para esas situaciones se usa el modelo funcional

```python
import tensorflow as tf 
import keras
from keras import layers
```

```python
#creando modelo secuencial 
model = keras.Sequential(
    [
        layers.Dense(2, activation='relu'),
        layers.Dense(3, activation='relu'),
        layers.Dense(4)
    ]
)

#mostrando los layers del modelo
model.layers
```

```python
#tambien se pueden crear el modelo agregando los layers o capas con la funcion add()

model = keras.Sequential()

model.add(layers.Dense(2, activation='relu'))
model.add(layers.Dense(3, activation='relu'))
model.add(layers.Dense(4))

model.layers
```

```python
#eliminar un layer con la funcion pop
model.pop()
print(len(model.layers))
```

```python
#nombrando los layers 
model = keras.Sequential(name="Mi_modelo_secuencial")

model.add(layers.Dense(2, activation="relu", name="layer1"))
model.add(layers.Dense(3, activation="relu", name="layer2"))
model.add(layers.Dense(4, name="Layer3"))

model.layers
```

```python
#especificando la forma de entrada 
layer = layers.Dense(3)
layer.weights
```

```python
x = tf.ones((1,4))
y = layer(x)
layer.weights
```

```python
model = keras.Sequential(
    [
        layers.Dense(2, activation="relu"),
        layers.Dense(3, activation="relu"),
        layers.Dense(4)
,    ]
)

x = tf.ones((1,4))
y = model(x)

print(f"Numero de pesos despues de llamar al modelo: {len(model.weights)}")
```
